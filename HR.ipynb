```python
import tensorflow as tf
from tensorflow.keras.applications import VGG16
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.models import Model
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
# Load pre-trained VGG16 model (without top layers)
base_model = VGG16(weights='imagenet', include_top=False, input_shape=(224, 224, 3))
```

```python
# Freeze base layers
for layer in base_model.layers:
    layer.trainable = False
```

```python
# Add custom layers for HR classification
x = Flatten()(base_model.output)
x = Dense(128, activation='relu')(x)
x = Dense(1, activation='sigmoid')(x)  # Binary Classification (HR vs Non-HR)
```

```python
# Compile model
model = Model(inputs=base_model.input, outputs=x)
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
# Load data (Assume train_data is preprocessed)
train_datagen = ImageDataGenerator(rescale=1./255)
train_data = train_datagen.flow_from_directory("E:\\Jupyter projects\\Datasets\\DRIVE\\DRIVE\\training", target_size=(224,224), batch_size=32, class_mode='binary')
```

```python
# Train model
model.fit(train_data, epochs=10)
model.save("hr_detection_model.h5")  # Save model
```
