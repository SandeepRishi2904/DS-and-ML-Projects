Advertisement Sales Prediction
1.FInding the Problem Application - whether a customer will buy the product or not
2.Collecting Dataset
3.Summarize the dataset - pandas library
4.Segregatting data into x and y - use iloc function --- x=dataset.iloc[:,:-1].values,y=dataset.iloc[:,-1].values --- iloc - integer location
5.Splitting data into train and test - train_test_split(x,y,test_size=0.25,random_state=42) --- 5/80% training and 25/20% testing   
6.Feature Scaling - Standardization and Normalization/Min-Max Scaling --- there will no bias
7.Algorithm - Logistic Regression
8.Training - model.fit(x_train,y_train)
9.Validation - data accuracy
10.Prediction - model.predict

```python
import numpy as np
import pandas as pd
```

```python
dataset=pd.read_csv()
```

```python
dataset.shape
```

```python
x=dataset.iloc[:,:-1].values #except last column all are stored in it
y=dataset.iloc[:,-1].values # last column is stored in it
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
```

```python
x_train=sc.fit_transform(x_train)
y_train=sc.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
print(np.concatenate((y_pred.reshape(len(y_pred),1),y_test.reshape(len(y_test),1)),1))
```

```python
from sklearn.metrics import accuracy_score
print("Accuracy Score: {0}%".format(accuracy_score(y_test,y_pred)*100))
```
